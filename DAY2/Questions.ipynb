```python
# import pyspark libraries 
# pyspark.sql :- main spark SQL module for dataframe operation
# pyspark.sql.functions :- built in function for df transfromataion (avg, max etc)
# pyspark.sql.types :- data types for defining schema ( StringType , Interger Type)
# pyspark.sql.window :- window function for advanced analytics
import pyspark.sql as ps
import pyspark.sql.functions as psf
import pyspark.sql.types as pst
import pandas as pd
import pyspark.sql.window as psw
```

# Filtering & Selection

6. Filter Rows where brand is "Abarth".

```python
# filter() or where() :- filteres rows based on a condition
# psf.expr() -> allows writing sql expressions as string 
# select :- for specific columns
# show :- display the results

(
     df
     .filter(psf.expr("brand=='Abarth'"))
     .select("brand","model")
     .show(10)
)
```

7. Find all vehicles with top_speed_kmh greater than 180.

```python
# query :- store a sql query as a multi line string
# df_view :- df_view is the temporary view created earlier
query = """
    select brand , model , top_speed_kmh
    from df_view
    where top_speed_kmh > 180
"""
```

# Aggregate Questions

11. Calculate the average efficiency_wh_per_km for each/per brand.

```python
# (
#     df.groupBy("brand").avg("efficiency_wh_per_km").show(10)
# )

(
    df
    .groupBy("brand")
    .agg(psf.avg("efficiency_wh_per_km")
    .alias("average"))
    .show(10)
)
```

12. Find the maximum top_speed_kmh for each brand.

```python
# In ascending order
# (
#     df.select("brand","top_speed_kmh").groupBy("brand")
#     .agg(psf.max("top_speed_kmh")
#     .alias("max_speed"))
#     .orderBy("max_speed")
#     .show(10)
# )

# in Descending order
(
    df.select("brand","top_speed_kmh").groupBy("brand")
    .agg(psf.max("top_speed_kmh")
    .alias("max_speed"))
    .orderBy(psf.col("max_speed").desc())
    .show(10)
)

```

# Handling Nulls

18. Count how many null values are present in each column.

```python
# psf.sum() -> sums all the null counts
# psf.col(c).isNull().cast("int") -> converts null check to 1 (null) or 0 (not null)
# transpose() -> transposes the result for easier viewing
(
    df
    .select([psf.sum(psf.col(c).isNull().cast("int")).alias(c) for c in df.columns])
    .show(20)
)
```

21. Fill nulls in towing_capacity_kg with the mean value of column.

```python
# psf.mean() -> calculate the average (mean) value
# first() -> gets the first (and only) row of the result
# [0] -> extract the value from the raw object.

mean_value = (
    df
    .select(psf.mean("towing_capacity_kg"))
    .first()
)[0]
```

```python
df_filled_with_mean = (
    df
    .select("brand","model","towing_capacity_kg")
    .fillna({"towing_capacity_kg":mean_value})
)

df_filled_with_mean.show(150)
```
