# Basic UDFs

28. create a UDF to categorize cars as "High range" (>300 km) or "Low Range"

```python
# Define a pandas UDF (user Defined function)
# taken a pandas series as input and return a pandas series
def categories_cars(range_km:pd.Series) -> pd.Series:
    return range_km.apply(lambda x: "High range" if x>300 else "low range")
```

```python
# register the pandas UDF with pyspark.
# psf.pandas_udf :- register the function as a pandas udf
# returnType :- specifies the retrun data type of the udf
categorize_udf = psf.pandas_udf(categories_cars,returnType=pst.StringType())


# apply the UDF to creare a new column 
# categorize_udf :- applies the UDF to the range_km column
# The UDF categorizes each row based on the range_km value 
df_with_Categories = (
    df.withColumn("Category",categorize_udf(psf.col("range_km")))
)


(
    df_with_Categories.select("brand","range_km","Category").show(10)
)
```

29. Create a UDF to Compute power_density = batter_capacity_kwh / length_mm.     -> kwh- kilowatt-hour -> 1kwh - 1 unit 

```python

# Define a pandas UDF with mulitple input columns 
def compute_power_density(capacity:pd.Series,length:pd.Series) -> pd.Series:
    result = capacity * length
    return result
```

```python
# Register the multi- inpout panad udf
compute_power_density_udf = psf.pandas_udf(compute_power_density,returnType = pst.FloatType())


# Applay the multi-input UDF to create a new column
df_with_power_density = (
    df.withColumn("power_density",compute_power_density_udf(psf.col("battery_capacity_kwh"),psf.col("length_mm")))
)


# Display the dataframe with the power density column
(
    df_with_power_density.select("brand","battery_capacity_kwh","length_mm","power_density").show(10)
)
```

42. Cache the dataframe and peform a groupby operation . Measure execution time before and after caching.

```python
import time
# without using cache 

start = time.time() # records the start time => 7.1
df.groupBy("car_body_type").count().show()

print("Without cache : ", time.time() - start) # 7.4 - 7.1 = 3 seconds 


df.cache() 

start = time.time()

df.groupBy("car_body_type").count().show()
print("With cache :", time.time() - start)
```

```python
# cache() -> store the dataframe in memory for faster execution
df.cache() 

start = time.time()

df.groupBy("car_body_type").count().show()
print("With cache :", time.time() - start)
```
